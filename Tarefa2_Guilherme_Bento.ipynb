```python
# -------------------- 0. Importing Packages --------------------

import matplotlib.pyplot as plt
import random
import math

# -------------------- 1. Calculating Pi with Monte Carlo --------------------

def pi_monteCarlo(N, plot=False, verbose=True):
    if plot:
        fig, ax = plt.subplots()
        fig.set_size_inches(4, 4)
        circle = plt.Circle((0, 0), 1, color='lightblue')
        square = plt.Rectangle((-1, -1), 2, 2, color='red', alpha=0.7)
        ax.add_patch(square)
        ax.add_patch(circle)

    # Creating parameters
    N_in = 0  # Number of dots in the circle
    cont = 0

    # Simulation
    while (cont < N):  # Number of interactions
        x = 2 * random.random() - 1  # Random number between 0 and 1 for the x
        y = 2 * random.random() - 1  # Random number between 0 and 1 for the y
        z = x**2 + y**2              # Checking if the dot is in the circle
        # Counting the number of dots inside the circle
        if (z <= 1):
            N_in += 1
            if plot:
                ax.plot(x, y, 'o', color='blue')
        else:
            if plot:
                ax.plot(x, y, 'o', color='pink')
        cont += 1

    # Estimating the value of pi
    pi_mc = 4 * N_in / N

    if verbose:  # Showing the text with the results
        print("Number of dots inside: ", N_in)
        print("Number of dots outside: ", N - N_in)
        print("Estimated Value de pi: ", pi_mc)
        print("Value de pi: ", math.pi)
        print("Error: ", pi_mc - math.pi)
        print("Precision: ", 100 * (1 - abs(pi_mc - math.pi)), "%")
        print("Number of Iterations: ", cont)

    if plot:
        plt.show()
    return pi_mc
```

```python
# -------------------- 0. Importing Packages --------------------

import matplotlib.pyplot as plt
import math

# -------------------- 1. Calculating Pi with Monte Carlo --------------------

val_N = [100, 200, 500, 1000, 2000, 5000, 10000] # Using different values of dots
simulations = 100
data = []

for N in val_N:
    results = []
    for _ in range(simulations):
        pi = pi_monteCarlo(N, plot=False, verbose=False)
        results.append(pi)

    data.append(results)

# Analysing the convergencing of the method
fig, ax = plt.subplots(figsize=(8, 6))
ax.boxplot(data, tick_labels=val_N)
ax.axhline(y=math.pi, color='red', linestyle='--', label='Pi (3.1415...)')
ax.set_title('Convergence of Monte Carlo Method for Pi')
ax.set_xlabel('Number of Dots (N)')
ax.set_ylabel('Estimated Value of Pi')
ax.legend()
plt.show()

# From the plot, we see thar increasing N reduces the variance of the estimates,
# demonstrating the convergence of the Monte Carlo method as statistical uncertainty decreases
```

```python
# -------------------- 0. Importing Packages --------------------

import matplotlib.pyplot as plt
import random

# -------------------- 1. Calculatin Pi with Monte Carlo --------------------
# --------------------    and Plooting the Graphs        --------------------

val_N = [500, 1000, 5000, 10000]    # Number of dots used in each simulation
fig, axs = plt.subplots(2, 2, figsize=(8, 8))
axs = axs.flatten()   # Transform the matriz 2x2 in a 1D list for the loop

for i in range(4):
    N = val_N[i]
    ax = axs[i]

    # Establishing tha parameters
    N_in = 0   # Number of dots inside the circle
    cont = 0

    # Simulating
    while cont < N:
        x = random.random()
        y = random.random()

        if (x-0.5)**2 + (y-0.5)**2 <= 0.5**2:
            N_in += 1
            ax.plot(x, y, 'o', color='black', markersize=2)
        else:
            ax.plot(x, y, 'o', color='red', markersize=2)
        cont += 1

    # Plotting
    ax.set_xlim(0, 1)
    ax.set_ylim(0, 1)
    ax.set_aspect('equal')
    ax.set_title(f"N = {N} | Pi Estimated Value = {4 * N_in / N:.4f}")

plt.tight_layout()
plt.show()

# We say as the number of dots increase the estimated valou of Pi converge more
# to his real value
```

```python
# -------------------- 0. Importing Packages --------------------

import matplotlib.pyplot as plt
import random
import math
import pandas as pd

# -------------------- 0. Importing Packages --------------------

val_N = [500, 1000, 5000, 10000]    # Number of dots used in each simulation
results = []

for i in range(4):
    N = val_N[i]

    # Establising tha parameters
    N_in = 0   # Number of dots inside the circle
    cont = 0

    # Simulating
    while cont < N:
        x = random.random()
        y = random.random()

        if (x-0.5)**2 + (y-0.5)**2 <= 0.5**2:
            N_in += 1
        cont += 1

    pi = 4 * N_in / N
    abs_error = abs(pi - math.pi)
    rel_error = abs_error / math.pi
    results.append({"N": N, "N_in": N_in,"Estimated Pi": pi,
                    "Absolute Error": abs_error, "%Error": rel_error})

# Creating the table with teh results
tabela = pd.DataFrame(results).round(4)
display(tabela)

# Again, with the increasing of dots used in Monte Carlo simulation the error
# decreases, indicating a better convergence of the value of Pi
```

```python
# -------------------- 0. Importing Packages --------------------

import random
import matplotlib.pyplot as plt

# -------------------- 1. Simulating a Random Walk --------------------

def randomwalk(x, time):
    Ts = [0]
    X = [0]
    for ts in range(1, time+1):
        Ts.append(ts)
        rand = random.choice(x)
        step = X[ts-1] + rand
        X.append(step)

    plt.figure(figsize = (8,6))
    plt.plot(Ts, X, label = 'Random walk')
    plt.xlabel('Time')
    plt.ylabel('Walk')
    plt.grid(True)
    plt.show()

randomwalk([-1, 1], 50)
```

```python
# -------------------- 0. Importing Packages --------------------

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm, binom

# -------------------- 1. implementing the Simulation --------------------

np.random.seed(1) # Defining a seed for the reproducibility of the results

# Parameters
times = [5, 500]
ensemble = 10000
results = {5: [], 500: []} # Defining a historical data to the times values

# Simulating random walk
for t in times:
    for i in range(ensemble):
        steps = np.random.choice([-1, 1], size=t)
        final_position = np.sum(steps)
        results[t].append(final_position)

# -------------------- 2. Plotting the Graphs --------------------

# Plotting the histograms
fig, axs = plt.subplots(1, 2, figsize=(10, 5))
for i, t in enumerate(times):
    ax = axs[i]

    # Some parameters to centralize the bins
    min_val = np.min(results[t])
    max_val = np.max(results[t])
    limites_bins = np.arange(min_val - 1, max_val + 2, 2)

    ax.hist(results[t], bins=limites_bins, density=True, alpha=0.7, color='blue', edgecolor='black')
    ax.set_title(f"Distribution for t={t}")
    ax.set_xlabel("Final Position")
    ax.set_ylabel("Density")

    if t == 5:
        # Binomial Distribution
        k = np.arange(0, t + 1)
        x_binom = 2 * k - t
        y_binom = binom.pmf(k, t, 0.5) / 2
        ax.plot(x_binom, y_binom, 'ro-', lw=2, label='Binomial')
    elif t == 500:
        # Gaussian Distribution
        x_gauss = np.linspace(-100, 100, 1001)
        y_gauss = norm.pdf(x_gauss, loc=0, scale=np.sqrt(t))
        ax.plot(x_gauss, y_gauss, color='red', lw=2, label='Gaussian')

    ax.legend()

plt.tight_layout()
plt.show()

# For t = 5, the final positions follow a discrete binomial distribution.
# For t = 500, since the final position is the sum of 500 independent and
# identically distributed steps, the distribution converges to a Gaussian
# distribution, showing the Central Limit Theorem result.
```

```python
# -------------------- 0. Importing Packages --------------------

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import skew, kurtosis

# -------------------- 1. Calcutating and Plotting --------------------
# --------------------    Statistical Moments      --------------------

np.random.seed(1) # Defining a seed for the reproducibility of the results

# Parameters
t_max = 1000
ensembles = [10, 100, 1000]
t = np.arange(1, t_max + 1)

# Preparing the plots
fig, axs = plt.subplots(2, 2, figsize=(10, 8))
axs = axs.flatten()
titles = ['Media', 'Variance', 'Skewness', 'Kurtosis']

# Generating a matrix with the steps in each ensemble
for M in ensembles:
    steps = np.random.choice([-1, 1], size=(M, t_max))

    # Calculating statiscal moments
    X = np.cumsum(steps, axis=1)
    media = np.mean(X, axis=0)
    variance = np.var(X, axis=0)
    assimetria = skew(X, axis=0)
    curtose = kurtosis(X, axis=0)

    # Plotting the curves
    axs[0].plot(t, media, label=f'M={M}', alpha=0.8)
    axs[1].plot(t, variance, label=f'M={M}', alpha=0.8)
    axs[2].plot(t, assimetria, label=f'M={M}', alpha=0.8)
    axs[3].plot(t, curtose, label=f'M={M}', alpha=0.8)

# Final details of the plots
for i, ax in enumerate(axs):
    ax.set_title(titles[i])
    ax.set_xlabel('Time (t)')
    ax.legend()
    ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

# As M increases, statistical fluctuations tend to vanish, indicating that
# with more simulations the sample moments converge to their expected
# theoretical values (mean = 0, variance = t, skewness = 0, kurtosis -> 0).
# This behavior is explained by the Law of Large Numbers: as the number of
# independent trials grows, the sample estimates approach the true expectation values.
#
# Note: In this code, kurtosis uses Fisher's definition (fisher=True), so the
# expected value for a Gaussian distribution is 0.
```
